```sql
SELECT 
    Id AS IdEmployee,
    EmployeeName,
    UPPER(JobTitle) AS JobTitle,
    TRY_CAST(COALESCE(NULLIF(BasePay, 'Not Provided'), '0.0') AS DOUBLE) AS BasePay,
    TRY_CAST(COALESCE(NULLIF(OvertimePay, 'Not Provided'), '0.0') AS DOUBLE) AS OvertimePay,
    TRY_CAST(COALESCE(NULLIF(OtherPay, 'Not Provided'), '0.0') AS DOUBLE) AS OtherPay,
    TRY_CAST(COALESCE(NULLIF(Benefits, 'Not Provided'), '0.0') AS DOUBLE) AS Benefits,
    CAST(TotalPay AS DOUBLE) AS TotalPay,
    CAST(TotalPayBenefits AS DOUBLE) AS TotalPayBenefits,
    CAST(Year AS INT) AS Year
FROM bronze.sf_salaries.salaries
WHERE EmployeeName != 'Not Provided' AND TotalPayBenefits >= 0.0;
```
