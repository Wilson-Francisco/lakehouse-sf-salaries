```sql
-- Verificar a estrutura da tabela
DESCRIBE TABLE bronze.sf_salaries.salaries;
```

```sql
-- Consulta de validação rápida para garantir a presença dos dados
SELECT 
    Year, 
    COUNT(*) as total_funcionarios,
    ROUND(AVG(TotalPayBenefits), 2) as media_salarial_total
FROM bronze.sf_salaries.salaries
GROUP BY Year
ORDER BY Year;
```
